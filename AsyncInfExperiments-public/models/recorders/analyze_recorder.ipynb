```python
from aiexperiments.visualizing.recording.recorder import Recorder
import matplotlib.pyplot as plt
import numpy as np
window_size = 5
def plot(axs, path):
    recorder = Recorder.load(path)
    # plot moving average
    moving_average_acc = np.convolve(recorder.accuracies, np.ones((window_size,))/window_size, mode='valid')
    axs[0].plot(moving_average_acc, label=path)
    moving_average_loss = np.convolve(recorder.losses, np.ones((window_size,))/window_size, mode='valid')
    axs[1].plot(moving_average_loss, label=path)
    print(path, np.mean(recorder.accuracies[-100:]))

fig, axs = plt.subplots(2, 1, figsize=(10, 10))
plot(axs, './dvsgesture/1.pickle')
axs[0].legend()
axs[0].set_title('Accuracy')
axs[1].set_title('Loss')
axs[0].grid()
axs[1].grid()
plt.show()
```
